## Loops en Python

```python
for numero in [0, 1, 2, 3, 4, 5]:
    print(numero)
```

```python
for letra in "Python":
    print(letra)
```

```python
paises = ["Argentina", "Colombia", "Perú", "México"]

for pais in paises:
    largo = len(pais)
    print(f"{pais} tiene {largo} letras")
```

```python
for pais in paises:
    if len(pais) > 6:
        print(f"{pais} tiene nombre largo")
    else:
        print(f"{pais} tiene nombre corto")
```

```python
colores = ["rojo", "azul", "verde"]
prendas = ["sombrero", "pantalon", "zapato"]

for color in colores:
    for prenda in prendas:
        print(prenda, color)
```
