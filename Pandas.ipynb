# Pandas

```python
# Pandas es una biblioteca de software libre escrita para el lenguaje de programación
# Python, diseñada específicamente para la manipulación y el análisis de datos.
# Proporciona estructuras de datos de alto rendimiento y herramientas de análisis fáciles de usar.
import pandas as pd
```

```python
# Datos representados en un diccionario de listas
datos = {"nombre": ["Pedro", "Juan", "Lorena"],
         "edad": [25, 39, 33]}

print(datos)
```

```python
# Un DataFrame es la estructura de datos principal de Pandas. Es una tabla bidimensional
# de datos etiquetados en filas y columnas, equivalente a una hoja de cálculo de Excel
# o a una tabla de base de datos relacional (SQL).
# Creación del Dataframe
df = pd.DataFrame(datos)

print(df)
```

```python
# Verificación del tipo de objeto del elemento completo 'df'
# Confirma que la variable es una estructura bidimensional de tipo DataFrame
# (pandas.core.frame.DataFrame)
type(df)
```

```python
# Acceso a una columna mediante corchetes
df["nombre"]
```

```python
# Acceso a una columna como atributo del DataFrame
df.nombre
```

```python
# Verificación del tipo de objeto:
# extrae una única columna devuelve una Pandas Series
type(df.nombre)
```

## Limpieza de datos

```python
# La limpieza de datos (data cleaning) en Pandas es el proceso de detectar, corregir
# o eliminar errores, inconsistencias, valores nulos o duplicados dentro de un conjunto
# de datos (Dataframe) para garantizar que los datos seanprecisos, completos y listos
# para análisis o aprendizaje automático.
data = {"id_producto": [1001, 1002, 1003, 1003],
        "cantidad_vendida": [30, None, 25, 25],
        "precio": [20.5, 15.0, None, 22.5]}
```

```python
df = pd.DataFrame(data)

# Mostrar los primeros 5 elementos
df.head()
```

```python
# Muestra un resumen general del DataFrame: tipos de datos, total de filas, columnas y memoria utilizada.
# Permite identificar de un vistazo cuántos valores nulos hay en cada columna
# (comparando 'non-null' contra el total de entradas)
df.info()
```

```python
# Evalúa cada celda del DataFrame para detectar valores nulos o faltantes (NaN/None)
# Devuelve una máscara booleana (True si el valor es nulo, False si contiene datos)
df.isnull()
```

```python
# # Cuenta la cantidad total de valores nulos o faltantes (NaN/None) presentes en cada columna del DataFrame
# Suma los valores booleanos True (1) generados por .isnull() para dar el conteo final por columna
df.isnull().sum()
```
