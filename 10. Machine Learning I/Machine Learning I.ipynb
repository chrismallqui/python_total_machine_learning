# **Regresión Lineal**

```python
# Importaciones 
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
df = pd.DataFrame()
df["Area"] = [2600, 3000, 3200, 3600, 4000]
df["Precio"] = [550000, 565000, 610000, 680000, 725000]
```

```python
plt.scatter(df["Area"], df["Precio"])
```

```python
type(df[["Area"]])
```

```python
type(df["Area"])
```

```python
X = df[["Area"]]
y = df["Precio"]
```

```python
modelo = linear_model.LinearRegression()
modelo.fit(X, y)
```

```python
modelo.predict([[3300]])
```

```python
plt.scatter(df["Area"], df["Precio"])
plt.scatter(3300, modelo.predict([[3300]]), color="red", marker="o", s=100)
plt.plot(df["Area"], modelo.predict(X))
plt.show()
```
